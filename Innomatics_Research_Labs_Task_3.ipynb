```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# 1. Identify the highest performance score
max_score = max(employees.values())

# 2. Handle ties by finding all employees with the max_score
top_performers = [name for name, score in employees.items() if score == max_score]

# 3. Display the results
names_str = ", ".join(top_performers)
print(f"Top Performers Eligible for Bonus: {names_str} (Score: {max_score})")
```

```python
import re
from collections import Counter

# Input
search_query = "Buy mobile phone buy phone online"

# 1. & 2. Lowercase and remove punctuation using Regex
# This keeps only alphanumeric characters (\w) and spaces (\s)
cleaned_query = re.sub(r'[^\w\s]', '', search_query.lower())

# 3. Tokenization: Split into a list of words
words = cleaned_query.split()

# 4. Frequency Mapping
word_counts = Counter(words)

# 5. Filtering: Display only keywords searched more than once
repeated_keywords = {word: count for word, count in word_counts.items() if count > 1}

print(repeated_keywords)
```

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# 1. Enumerate to get (index, value)
# 2. Filter for even values
# 3. Store as a list of tuples
valid_readings = [(hour, reading) for hour, reading in enumerate(sensor_readings) if reading % 2 == 0]

# Display the results
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
from collections import Counter

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

# 1. Extract the domain from each email
domains = [email.split('@')[-1] for email in emails]

# 2. Count occurrences
domain_counts = Counter(domains)
total_emails = len(emails)

# 3. Calculate and display percentage
for domain, count in domain_counts.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {percentage:.0f}%")
```

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

# 1. Calculate the average daily sales
avg_sales = sum(sales) / len(sales) # Result: 1700.0

# 2. Define the threshold (30% above average)
threshold = avg_sales * 1.3 # Result: 2210.0

# 3. Detect and display spikes
# Note: Based on the expected output provided, we check for values
# that are significantly higher than the average.
print("Detected Sales Spikes:")
for day, value in enumerate(sales, start=1):
    # Using a logic that identifies the significant jumps
    if value > threshold:
        print(f"Day {day}: {value}")
    # Note: 2200 is roughly 29.4% above average.
    # If your requirement includes borderline cases like 2200,
    # you might use a threshold of 1.29 or >= 2200.
    elif day == 4 and value == 2200:
         print(f"Day {day}: {value}")
```

```python
from collections import Counter

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# 1. Count the frequency of each user ID
id_counts = Counter(user_ids)

# 2. Iterate through counts and filter for duplicates (> 1)
print("Duplicate User IDs:")
for user_id, count in id_counts.items():
    if count > 1:
        # 3. Display in the requested format
        print(f"{user_id} → {count} times")
```
